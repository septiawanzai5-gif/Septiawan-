## Langkah 10: Tantangan opsional, kata "tidak"
**Tidak wajib.** Jika dikerjakan, kamu mendapat nilai tambah (nilai maksimum tetap 100).

Pada fungsi di bawah, jika kata sebelum sebuah kata kunci adalah *tidak*, *bukan*, atau *kurang*, nilai skornya dibalik (positif menjadi negatif, dan sebaliknya).

```python
def tebak_dengan_tidak(kalimat):
    kata = kalimat.lower().replace(",", " ").replace(".", " ").split()
    skor = 0
    for i in range(len(kata)):
        nilai = 0
        if kata[i] in kata_positif:
            nilai = 1
        if kata[i] in kata_negatif:
            nilai = -1
        if i > 0 and kata[i - 1] in ["tidak", "bukan", "kurang"]:
            # TODO: balik nilai (kalikan dengan -1)
            pass
        skor = skor + nilai
    if skor > 0:
        return "positif"
    elif skor < 0:
        return "negatif"
    else:
        return "netral"

df["tebakan2"] = df["kalimat"].apply(tebak_dengan_tidak)
akurasi_tantangan = (df["tebakan2"] == df["label"]).mean() * 100
print("Akurasi dengan aturan tidak:", round(akurasi_tantangan, 1), "%")
```
